```python
import json
import numpy as np

# base_cam.jsonから回転行列と平行移動ベクトルを取得
with open(base_cam_path, 'r') as f:
    base_cam_data = json.load(f)

# 2.5の手順１の(3)式に対応
# カメラの中心座標を適切な座標系に変換して追加
with open(NeRF_mesh_camera_points_path, 'w') as ply_file:
    ply_file.write("ply\nformat ascii 1.0\n")
    ply_file.write("comment Added camera center coordinates\n")
    ply_file.write(f"element vertex {len(base_cam_data['path'])}\n")
    ply_file.write("property float x\nproperty float y\nproperty float z\n")
    ply_file.write("property uchar red\nproperty uchar green\nproperty uchar blue\n")
    ply_file.write("end_header\n")
    for i in range(len(base_cam_data['path'])):
        cx = (base_cam_data['path'][i]['T'][0] - 0.025 - 0.5)/0.33
        cy = (base_cam_data['path'][i]['T'][1] - 0.025 - 0.5)/0.33
        cz = (base_cam_data['path'][i]['T'][2] - 0.025 - 0.5)/0.33
        R = np.array([[0,1,0],[1,0,0],[0,0,-1]])
        x,y,z = np.dot(R,np.array([cx,cy,cz]))
        # plyファイルにカメラの中心座標を追加
        ply_file.write(f"{cx:.5f} {cy:.5f} {cz:.5f} 255 0 0\n")
```

# 2.5の手順2 colmapカメラ座標点群生成

```python
import numpy as np

import json

def parse_images_txt(file_path, output_file_path):
    # ファイルを読み込んで行ごとに処理
    with open(file_path, "r") as images_text:
        lines = images_text.readlines()

    # 画像情報を格納するリスト
    image_data = []

    for i in range(len(lines)):
        line = lines[i].strip()
        if line.startswith("#"):
            continue

        if i % 2 == 0:
            # 画像情報の行
            image_info = line.split(" ")
            image_id = int(image_info[0])
            qw, qx, qy, qz = map(float, image_info[1:5])
            tx, ty, tz = map(float, image_info[5:8])
            camera_id = int(image_info[8])
            name = image_info[9]

            current_image = {
                'IMAGE_ID': image_id,
                'QW': qw,
                'QX': qx,
                'QY': qy,
                'QZ': qz,
                'TX': tx,
                'TY': ty,
                'TZ': tz,
                'CAMERA_ID': camera_id,
                'NAME': name,
                'POINTS2D': []
            }
        else:
            # 画像の座標と特徴量情報
            feature_ls = line.split(" ")
            for j in range(2, len(feature_ls), 3):
                if feature_ls[j] != '-1':
                    x, y, point3d_id = map(float, feature_ls[j - 2:j + 1])
                    current_image['POINTS2D'].append((x, y, str(int(point3d_id))))

            # -1以外の画像情報をリストに追加
            if current_image['IMAGE_ID'] != -1:
                image_data.append(current_image)

    os.makedirs(os.path.join(os.getcwd(), 'json'), exist_ok=True)
    with open(os.path.join(os.getcwd(), 'json', output_file_path), 'w') as json_file:
        json.dump(image_data, json_file, indent=4)

    print(f"画像情報を {output_file_path} に保存しました。")

    return image_data

image_data = parse_images_txt(file_path=os.path.join(os.getcwd(), "colmap_text", "images.txt"), output_file_path="image_data.json")

# カメラの中心座標を適切な座標系に変換して追加
with open(os.path.join(os.getcwd(), "colmap_camera_points.ply"), 'w') as ply_file:
    ply_file.write("ply\nformat ascii 1.0\n")
    ply_file.write("comment Added camera center coordinates\n")
    ply_file.write(f"element vertex {len(base_cam_data['path'])}\n")
    ply_file.write("property float x\nproperty float y\nproperty float z\n")
    ply_file.write("property uchar red\nproperty uchar green\nproperty uchar blue\n")
    ply_file.write("end_header\n")

    for i in range(len(image_data)):
        # 画像毎の四元数と並進ベクトルを取り出す
        qw, qx, qy, qz, tx, ty, tz = image_data[i]['QW'], image_data[i]['QX'], image_data[i]['QY'], image_data[i]['QZ'], image_data[i]['TX'], image_data[i]['TY'], image_data[i]['TZ']

        # 四元数を正規化
        quaternion = np.array([qw, qx, qy, qz])
        quaternion /= np.linalg.norm(quaternion)

        # 四元数から回転行列を計算
        R = np.array([
            [1 - 2*qy*qy - 2*qz*qz, 2*qx*qy - 2*qz*qw, 2*qx*qz + 2*qy*qw],
            [2*qx*qy + 2*qz*qw, 1 - 2*qx*qx - 2*qz*qz, 2*qy*qz - 2*qx*qw],
            [2*qx*qz - 2*qy*qw, 2*qy*qz + 2*qx*qw, 1 - 2*qx*qx - 2*qy*qy]
        ])

        # カメラ座標から世界座標に変換する
        translation = np.array([tx, ty, tz])
        camera_center = -np.dot(np.transpose(R), translation)
        cx, cy, cz = camera_center
        ply_file.write(f"{cx:.5f} {cy:.5f} {cz:.5f} 255 0 0\n")
        # print(i+1, cx, cy, cz)
```

```python
import open3d as o3d
import numpy as np

# MVS点群とInstantNGP点群のパス
colmap_path = os.path.join(os.getcwd(), "combined_colmap.ply")
nerf_path = os.path.join(os.getcwd(), "combined_nerf.ply")

# MVS点群とInstantNGP点群を読み込む
colmap_mesh = o3d.io.read_triangle_mesh(colmap_path)
nerf_mesh = o3d.io.read_triangle_mesh(nerf_path)

# InstantNGP点群をMVS点群の座標系に移動
nerf_mesh_center = nerf_mesh.get_center()
colmap_mesh_center = colmap_mesh.get_center()
translation_vector = colmap_mesh_center - nerf_mesh_center
print(translation_vector)
nerf_mesh.translate(translation_vector)

# MVS点群のスケールを取得
colmap_mesh_size = np.linalg.norm(colmap_mesh.get_max_bound() - colmap_mesh.get_min_bound())
print(colmap_mesh.get_max_bound() - colmap_mesh.get_min_bound(), colmap_mesh_size)
# InstantNGP点群のサイズを取得
nerf_mesh_size = np.linalg.norm(nerf_mesh.get_max_bound() - nerf_mesh.get_min_bound())
print(nerf_mesh.get_max_bound() - nerf_mesh.get_min_bound(), nerf_mesh_size)
print((colmap_mesh.get_max_bound() - colmap_mesh.get_min_bound())/(nerf_mesh.get_max_bound() - nerf_mesh.get_min_bound()))
# InstantNGP点群をMVS点群と同じ大きさにスケール変換
scale_factor = colmap_mesh_size / nerf_mesh_size
nerf_mesh.scale(scale_factor, center=nerf_mesh.get_center())

# 結合されたメッシュを作成
mesh_combined = nerf_mesh

# 結合されたメッシュを保存
output_mesh_path = os.path.join(os.getcwd(), "scaled_combined_nerf.ply")
o3d.io.write_triangle_mesh(output_mesh_path, mesh_combined)
```
